```python
import pandas as pd
import numpy as np

#Defining intial 
x = np.array([[-1,1],[1,1],[-2,-1],[2,-1]])
x = x.T
y = np.array([1,1,0,0])

#Our parameters
theta = np.array([0.2,-0.1])
alpha = 1
m = len(y)

#Def sig
def sig(z):
    res = 1/(1+np.e**(-z))
    return res

#definight h as pred
def pred(x,theta):
    ans = theta @ x
    res = sig(ans)
    return res

#Theta
def update(x,y,h,theta):
    l = len(theta)
    grad_theta = np.zeros(l)
    for i in range(l):
        grad_theta[i] = (1/m)*np.sum((h-y)* x[i,:])
        theta[i] = theta[i] - alpha * grad_theta[i]
    return theta

def cost(x,y,h):
    res = -(1/m) * np.sum(y*np.log(h)+(1-y)*np.log(1-h))
    return res

print('iteration 0', theta)

def iteration(x,y,theta,num):
    
    for j in range(num):
        h = pred(x,theta)
        cos = cost(x, y, h)
        theta = update(x, y, h, theta)
        print('iteration',j+1,theta)
        print('cost is', cos)
    return theta

print(iteration(x,y,theta,20))


#this works perfectelly 


```

```python
#trying to represent results on a graph
#To express our line in terms of x1 and x2 we want to assume h(z) = 0.5 which gives ans = 0 sp we just rearange theta * x = 0
import matplotlib.pyplot as plt
import numpy as np



x = np.array([[-1,1],[1,1],[-1,-1],[1,-1]])
x = x.T
y = np.array([1,1,0,0])
theta = np.array([0.2,-0.1])

xp = np.linspace(-3,3,100)
theta = iteration(x,y,theta,20)
yp = -(theta[0]/theta[1]) * xp

dec = y >0.5 #We use dec to turn y into true or false statemtn

x_y = x.T[dec]
x_y = x_y.T

x_n = x.T[~dec]
x_n = x_n.T
print(x_y)
print(x_n)
#We use there varaible x_n and x_y to seperate between these two varaibles

plt.scatter(x_y[0,:], x_y[1,:], marker="x", color="red", label='y=1')
plt.scatter(x_n[0,:], x_n[1,:], marker="o", color="orange", label='y=0')
plt.plot(xp, yp, label='learned boandry')
plt.xlim(-1.5, 1.5)
plt.ylim(-1.5,1.5)
plt.xlabel('x0')
plt.ylabel('x1')
plt.title('PLot')
plt.grid(True)
plt.legend()


plt.show
```
