```python
from sklearn.datasets import load_breast_cancer
import pandas as pd
import numpy as np

data = load_breast_cancer()

x = data.data
y = data.target #Y is true value of y
print(x.shape)
print(y.shape)
# We want the 569 examples to be lenght of our matrix so we contiously run it through our theta "vector"
 #We opertaing with (m examples, n features)
#log regression
```

```python
#Creating theta and x
#Defining our parameters
theta = np.zeros(30)
theta[0] = 1
#print(theta)

alpha = 0.5

#Define sigmoid or logistic function
def sig(z):
    res = 1/(1+np.e**(-z))
    return res

#Def h(x)
def h(x,theta):
    #print(x.shape)
    z = theta @ x.T
    result = sig(z)#1/(1+(np.e)**(-z)) # This our sigmoid function as contained in h_theta (x)
    return result # Here as

#print(h(x,theta))

def classification(x):
    if x>= 0.5:
        res = 1
    elif x < 0.5:
        res = 0
    return res


#We assume P(y=1) = h_theta(x)
#Assume P(y=0) = 1- h_theta(x)
# P(y) = h(x)^y * (1-h(x))^(1-y)

#L(theta) = (multiplactive sum over i) h(x_i)^y_i * 1-h_xi)^1-yi
#Finding log loss function
#l(theta) =  log(L(theta)) 
        #= sum (y_i log(h(x_i))+ (1-y_i) log (1-h_theta(x_i))
#Choose theta to minimise
#Gradiet ascent gives plus instead ofminus
# theta_j := theta + d/dtheta_j  l(theta)
m=569

def cost(x,y,h,theta):
    #print('Theta', theta[0:2])
    h1 = h(x,theta)
    #print('h',h(x,theta)[0:2])
    #print('h1', h1[0:2])
    #print(h1[0])
    h1 = np.clip(h1, 1e-15, 1-1e-15) #clip h so we dont get errors
    #print(h1[0])
    sum1 = np.sum( y * np.log(h1) + (1-y)*np.log(1-h1)  )
    #print(sum1)
    result = -(1/m) * sum1
    #print('res', result)
    return result

#Combining these two equations gives 
#theta_j := theta_j + alpha sum(   (y_i - h(x_i)) * x_j  ) # Spaces are put here to show that its in the sum

def update(theta, h, x, y, alpha):
    #print(h(x,theta).shape)
    #print(y.shape)
    #print((x).shape)
    #print((x[0][:].shape))
    #h = np.clip(h, 1e-15, 1-1e-15) # We want to clip h so we dont get errors
    h1 = h(x,theta)
    su = np.sum(  (y - h1)[:, None] * x, axis=0  ) #[:, None] turns y-h into (569, 1) and axis = 0 unsures it sums over rows
    #print(su.shape)
    #print(su)
    #print("Before", theta)
    theta = theta + alpha * su
    #print("after", theta)
    return theta

#print("before",cost(x,y,h,theta))
#print(theta)
theta = update(theta, h, x, y, alpha)
#print("after",cost(x,y,h,theta))
#print(theta)
theta = update(theta, h, x, y, alpha)
#print("after",cost(x,y,h,theta))
#
```

```python
def learn(x,y,h,alpha, num):
    theta = np.zeros(30)
    for i in range(num):
        theta = update(theta, h, x, y, alpha)
        print("iteration", i)
        #print(theta)
        print(cost(x,y,h,theta))
        print('begin h',h(x,theta)[0:2])
        print('end of h',h(x,theta)[-2:0])
    return theta

learn(x,y, h, alpha, 10)
```
